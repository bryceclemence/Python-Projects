```python
import win32com.client
import time

report_pathway = 'report/pathway'
datestr = str(time.strftime("_%m.%d.%Y"))

# Start an instance of Excel
xlapp = win32com.client.dynamic.Dispatch("Excel.Application")

# Open the workbook in said instance of Excel
wb = xlapp.Workbooks.Open(report_pathway)

# Refresh all data connections twice to make sure pivot tables update as well.
wb.RefreshAll()
xlapp.CalculateUntilAsyncQueriesDone()
wb.RefreshAll()
xlapp.CalculateUntilAsyncQueriesDone()


# Save Report
xlapp.DisplayAlerts = False
wb.Save()
xlapp.DisplayAlerts = True

# Quit
xlapp.Quit()
xlapp=None

with open(report_pathway+".xlsx", "rb") as f1:
    with open(report_pathway + str(datestr)+".xlsx", "wb") as f2:
        f2.write(f1.read())
```
